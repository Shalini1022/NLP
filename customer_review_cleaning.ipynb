```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download NLTK data
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

# 1. Load IMDb dataset
df = pd.read_csv("IMDB_dataset.csv")

print("Dataset:")
print(df.head())

# 2. Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Check duplicate reviews
print("\nDuplicate reviews:", df['review'].duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# NLTK tools
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

# Cleaning function
def clean_text(text):

    # 4. Lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # 5. Remove numbers and special characters
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # 6 & 7. Tokenize and remove stopwords
    words = word_tokenize(text)
    words = [word for word in words if word not in stop_words]

    # 8. Stemming
    words = [stemmer.stem(word) for word in words]

    return ' '.join(words)

# Apply cleaning
df['cleaned_review'] = df['review'].apply(clean_text)

# 9. Compare original and cleaned reviews
print("\nOriginal Review:")
print(df['review'].iloc[0])

print("\nCleaned Review:")
print(df['cleaned_review'].iloc[0])

# 10. Export dataset
df.to_csv("IMDB_Cleaned.csv", index=False)

print("\nCleaned dataset saved as IMDB_Cleaned.csv")
```
