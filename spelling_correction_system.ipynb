```python
import re

# 1. Load spelling-error corpus
with open("spelling_corpus.csv", "r") as f:
    text = f.read().lower()

# 2. Build vocabulary of correctly spelled words
vocab = set(re.findall(r'\b[a-z]+\b', text))

# Edit Distance function
def edit_distance(a, b):
    if not a:
        return len(b)
    if not b:
        return len(a)

    if a[0] == b[0]:
        return edit_distance(a[1:], b[1:])

    return 1 + min(
        edit_distance(a[1:], b),
        edit_distance(a, b[1:]),
        edit_distance(a[1:], b[1:])
    )

# 3. Accept user query
query = input("Enter search query: ").lower()

# 4. Tokenize query
words = query.split()

corrected = []

# 5-7. Find incorrect words and closest word
for word in words:
    if word in vocab:
        corrected.append(word)
    else:
        best_word = min(vocab, key=lambda x: edit_distance(word, x))
        print("Incorrect:", word)
        print("Suggested:", best_word)
        corrected.append(best_word)

# 8. Display corrected query
print("\nOriginal query:", query)
print("Corrected query:", " ".join(corrected))
```
