```python
import nltk

# 1. Tokenize - split text
from nltk.tokenize import word_tokenize, sent_tokenize
text = input('enter text') //get input from user
print(word_tokenize(text)) //token text
print(sent_tokenize(text)) //input text

# 2. Stem - word reduction
from nltk.stem import PorterStemmer
stem = PorterStemmer()
print("\nStemming:")
print(stem.stem("playing"))


# 3. Lemma - meaningful root word
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()
print("\nLemmatization:")
print(lemma.lemmatize("cars"))


# 4. Tag - grammar labeling
from nltk import pos_tag
words = word_tokenize("Python is powerful")
print("\nPOS Tag:")
print(pos_tag(words))

# 5. Chunk - phrase extraction
from nltk.chunk import RegexpParser
chunk = RegexpParser("NP:{<DT>?<JJ>*<NN>}")
print("\nChunking:")
print(chunk.parse(pos_tag(word_tokenize("The big dog"))))


# 8. Classify - prediction
from nltk.classify import NaiveBayesClassifier

data = [
    ({"word":"good"}, "positive"),
    ({"word":"bad"}, "negative")
]

classifier = NaiveBayesClassifier.train(data)

print("\nClassification:")
print(classifier.classify({"word":"good"}))


# 9. Metrics - accuracy
from nltk.metrics import accuracy

print("\nMetrics:")
print(accuracy(["yes","no"],["yes","yes"]))


# 10. Probability - frequency
from nltk.probability import FreqDist

print("\nProbability:")
print(FreqDist(["python","python","nltk"]))


# 11. Semantics - meaning
from nltk.corpus import wordnet

print("\nSemantics:")
print(wordnet.synsets("car")[0].definition())


# 12. Util - helper tools
from nltk.util import bigrams

print("\nUtil:")
print(list(bigrams(["I","love","Python"])))
```
