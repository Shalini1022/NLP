```python
# Install NLTK
!pip install nltk


import nltk

# Download required packages
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')


from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.util import bigrams, trigrams
from nltk.probability import FreqDist
from nltk.classify import NaiveBayesClassifier
from nltk.metrics import accuracy
from nltk.sem import Expression
from nltk import CFG
from nltk.parse import ChartParser



# -------------------------------
# News Article Input
# -------------------------------

news = """
Microsoft announced a new AI platform in Chennai.
The company expects the technology to improve business productivity.
"""


# -------------------------------
# 1. Parsing using nltk.parse
# -------------------------------

print("----- Parsing -----")


grammar = CFG.fromstring("""
S -> NP VP
NP -> 'Microsoft'
VP -> V NP
V -> 'announced'
NP -> 'AI'
""")


parser = ChartParser(grammar)


sentence = ['Microsoft','announced','AI']


for tree in parser.parse(sentence):
    print(tree)


print("Subject : Microsoft")
print("Verb : announced")
print("Object : AI platform")



# -------------------------------
# 2. Named Entity Recognition using nltk.chunk
# -------------------------------


print("\n----- Named Entities -----")


tokens = word_tokenize(news)

tags = pos_tag(tokens)

named_entities = ne_chunk(tags)


print(named_entities)



# -------------------------------
# 3. Classification using nltk.classify
# -------------------------------


print("\n----- News Category -----")


training_data = [

({"AI":True, "technology":True}, "Technology"),

({"match":True, "player":True}, "Sports"),

({"election":True, "government":True}, "Politics"),

({"market":True, "business":True}, "Business")

]


classifier = NaiveBayesClassifier.train(training_data)



test_article = {

"AI":True,

"technology":True

}


result = classifier.classify(test_article)


print("Category :", result)




# -------------------------------
# 4. Accuracy using nltk.metrics
# -------------------------------


print("\n----- Model Accuracy -----")


test_data = [

({"AI":True,"technology":True},"Technology"),

({"match":True,"player":True},"Sports")

]


actual_values = []

predicted_values = []


for features,label in test_data:

    actual_values.append(label)

    predicted_values.append(classifier.classify(features))


model_accuracy = accuracy(actual_values,predicted_values)


print("Accuracy :", model_accuracy*100,"%")




# -------------------------------
# 5. Word Frequency and Probability
# -------------------------------


print("\n----- Word Frequency -----")


words = word_tokenize(news.lower())


frequency = FreqDist(words)



print("AI :", frequency["ai"])

print("technology :", frequency["technology"])

print("business :", frequency["business"])



print("\n----- Probability -----")


total = len(words)


probability = frequency["ai"] / total


print("P(AI) =",
      frequency["ai"],
      "/",
      total,
      "=",
      probability)




# -------------------------------
# 6. Semantic Analysis using nltk.sem
# -------------------------------


print("\n----- Semantic Meaning -----")


expression = Expression.fromstring("technology(x)")


print(expression)




# -------------------------------
# 7. Bigrams and Trigrams
# -------------------------------


print("\n----- Bigrams -----")


bi = list(bigrams(words))


for item in bi:

    print(item)



print("\n----- Trigrams -----")


tri = list(trigrams(words))


for item in tri:

    print(item)
```
