```python

import re
import math
import numpy as np
import nltk
from collections import Counter
from datasets import load_dataset

print("Loading WikiText-2 dataset...")

dataset = load_dataset(
    "Salesforce/wikitext",
    "wikitext-2-raw-v1"
)

print("Dataset loaded successfully!")


text = " ".join(dataset["train"]["text"])

text = text.lower()

text = re.sub(
    r"[^a-z\s]",
    " ",
    text
)

text = re.sub(
    r"\s+",
    " ",
    text
).strip()


tokens = text.split()

print("Total tokens:", len(tokens))

# Use 100,000 tokens for faster processing
tokens = tokens[:100000]

print(
    "Tokens used:",
    len(tokens)
)


unigrams = Counter(tokens)

bigrams = Counter(
    zip(
        tokens[:-1],
        tokens[1:]
    )
)

trigrams = Counter(
    zip(
        tokens[:-2],
        tokens[1:-1],
        tokens[2:]
    )
)

print(
    "Vocabulary size:",
    len(unigrams)
)

print(
    "Unique bigrams:",
    len(bigrams)
)

print(
    "Unique trigrams:",
    len(trigrams)
)


def predict_next_word(sentence, top_n=5):

    sentence = sentence.lower()

    sentence = re.sub(
        r"[^a-z\s]",
        " ",
        sentence
    )

    words = sentence.split()

    if not words:
        return []


    predictions = {}


    # Trigram prediction
    if len(words) >= 2:

        word1 = words[-2]
        word2 = words[-1]

        total = sum(
            count
            for (w1, w2, w3), count
            in trigrams.items()
            if w1 == word1 and w2 == word2
        )


        if total > 0:

            for (
                w1,
                w2,
                w3
            ), count in trigrams.items():

                if (
                    w1 == word1
                    and w2 == word2
                ):

                    predictions[w3] = (
                        count / total
                    )


    # Bigram fallback
    if not predictions:

        previous_word = words[-1]

        total = unigrams[
            previous_word
        ]


        if total > 0:

            for (
                w1,
                w2
            ), count in bigrams.items():

                if w1 == previous_word:

                    predictions[w2] = (
                        count / total
                    )


    # Sort by probability
    result = sorted(
        predictions.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return result[:top_n]

print("\n" + "=" * 50)

print(
    "       SMART NEXT-WORD PREDICTOR"
)

print("=" * 50)

print(
    "Application: Mobile Keyboard / Chat Autocomplete"
)

print(
    "Type 'exit' to stop."
)


while True:

    user_input = input(
        "\nEnter a sentence: "
    )


    if user_input.lower() == "exit":

        print(
            "Program stopped."
        )

        break


    predictions = predict_next_word(
        user_input,
        top_n=5
    )


    if predictions:

        print(
            "\nTop Next-Word Predictions:"
        )

        for i, (
            word,
            probability
        ) in enumerate(
            predictions,
            1
        ):

            print(
                f"{i}. {word} "
                f"- Probability: "
                f"{probability:.4f}"
            )

    else:

        print(
            "No prediction available."
        )
```
